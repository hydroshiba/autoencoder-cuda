```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("training_log.csv")

plt.figure()
plt.plot(data["epoch"], data["loss"], marker="o")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss")
plt.grid()
plt.savefig("loss_curve.png")

plt.figure()
plt.plot(data["epoch"], data["time_ms"], marker="o")
plt.xlabel("Epoch")
plt.ylabel("Time (ms)")
plt.title("Training Time per Epoch")
plt.grid()
plt.savefig("time_curve.png")

plt.show()
```
